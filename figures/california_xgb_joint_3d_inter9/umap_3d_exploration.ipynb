# UMAP 3D – exploration interactive

Ce notebook charge les coordonnées UMAP 3D calculées pour les 9 interactions les plus fortes
(`joint_shap_4clusters_inter9_3d.csv`) et propose des visualisations interactives (Plotly)
et Matplotlib 3D pour explorer les 4 clusters et la distribution du prix.

```python
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

# Charger le CSV généré par le pipeline 3D
csv_path = "joint_shap_4clusters_inter9_3d.csv"
df = pd.read_csv(csv_path)

print(df.shape)
df.head()
```

```python
# Vue 3D interactive avec Plotly – couleur = cluster
fig_cluster = px.scatter_3d(
    df,
    x="umap_1_3d",
    y="umap_2_3d",
    z="umap_3_3d",
    color="cluster_inter9_4_3d",
    hover_data=["price", "Latitude", "Longitude"],
    title="UMAP 3D – 4 clusters (interactions top9)",
)
fig_cluster.show()
```

```python
# Vue 3D interactive avec Plotly – couleur = prix
fig_price = px.scatter_3d(
    df,
    x="umap_1_3d",
    y="umap_2_3d",
    z="umap_3_3d",
    color="price",
    color_continuous_scale="Viridis",
    hover_data=["cluster_inter9_4_3d", "Latitude", "Longitude"],
    title="UMAP 3D – couleur = prix",
)
fig_price.show()
```

```python
%matplotlib notebook

# Vue 3D Matplotlib (interactive dans Jupyter)
x = df["umap_1_3d"].values
y = df["umap_2_3d"].values
z = df["umap_3_3d"].values
labels = df["cluster_inter9_4_3d"].values

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection="3d")
for c in sorted(set(labels)):
    m = labels == c
    ax.scatter(x[m], y[m], z[m], s=6, alpha=0.7, label=f"Cluster {int(c)}")
ax.set_title("UMAP 3D – 4 clusters (interactions top9)")
ax.set_xlabel("UMAP 1")
ax.set_ylabel("UMAP 2")
ax.set_zlabel("UMAP 3")
ax.legend(markerscale=3, fontsize=8)
plt.show()
```
